```python
dp = [[] for _ in range(N)]
dp[0] = [[1<<0, 0, 0]]
```

```python


for n in range(1, N):
    appended = False
    for hist, cur_value, cur_loc in dp[n-1]:
        for sell_loc in range(N):
            sell_value = arr[cur_loc][sell_loc]
            if sell_value >= cur_value and not ( (1 << sell_loc) & hist):
                appended = True
                dp[n].append( [hist | (1<sell_loc), sell_value, sell_loc])
    if not appended:
        break

for i in range(N-1, -1, -1):
    if len(dp[i]) != 0:
        print(i+1)
        break
```

```python
dp = [ [11 for _ in range((1<<(N)))] for _ in range(N)]
dp[0][1<<0] = 0


for n in range(1, N):
    changed = False
    for cur_loc in range(N):
        for hist in range((1<<(n)) -1, (1<<(n+1)) - 1): ### 히스토리들을 볼거야
            if dp[cur_loc][hist] < 11: ### 바꼈다면!
                for next_loc in range(N): ### arr의 보기에서 골라올 next_loc
                    if arr[cur_loc][next_loc] >= dp[cur_loc][hist] and not (1 << next_loc) & hist:
                        changed = True
                        dp[next_loc][hist | (1<<next_loc)] = min(arr[cur_loc][next_loc], dp[next_loc][hist | (1<<next_loc)])
    if not changed:
        n -= 1
        break

print(n+1)

```
