```python

INF = float('inf')
```

```python
N = 5
arr =  list(map(int, "-2 4 -99 -1 98".split()))
arr.sort()
```

```python
start = 0
end = N - 1

while start <= end:
    mid = (start + end) // 2

    print(start, end, mid)

    if arr[mid] < 0:
        start = mid + 1
    elif arr[mid]> 0:
        end = mid - 1

acid = mid
```

```python
start = 0
end = N - 1

ans = [INF, -1, -1]
while start < acid and end >= acid:

    cur_al = arr[start]
    cur_ac = arr[end]

    mal = cur_al + cur_ac
    ans = min(ans, [abs(mal), cur_al, cur_ac])
    if mal < 0:
        start +=1
    elif mal > 0:
        end -=1
    else:
        break
print(" ".join(map(str, ans[1:])))
```
