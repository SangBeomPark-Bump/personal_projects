```python
N = 5

raw_arr = """3 7 2 0 1
2 8 0 9 1
1 2 1 8 1
9 8 9 2 0
3 6 5 1 5""".split('\n')

arr =  [list(map(int,row.split())) for row in raw_arr]
prob_num = 2
```

```python
from heapq import heappop, heappush
```

```python

INF = float("INF")
arr2 = [ [INF for _ in range(N)] for _ in range(N)]
arr2[0][0] = arr[0][0]

dx = (0, 0,1,  -1)
dy = (1, -1, 0, 0)
que = []
heappush(que, (arr[0][0], (0, 0)))
cnt = 0
while que:
    dist, (y, x) =  heappop(que)
    for i in range(4):
        ny = y + dy[i]
        nx = x + dx[i]
        if not(0 <= ny < N and 0 <= nx < N):
            continue
        else:
            next_dist = dist + arr[ny][nx]
\
            if next_dist < arr2[ny][nx]:
                cnt +=1
                arr2[ny][nx] = next_dist
                heappush(que, (next_dist,(ny, nx)))
                
print(f'Problem {prob_num}: {arr2[N-1][N-1]}')
```
