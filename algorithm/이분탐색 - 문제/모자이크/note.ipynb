```python
N, M = 4, 14
P = 4
L = 9

raw_arr = """1 2
2 1
2 3
1 6
3 5
1 10
3 6
1 12
2 13""".split('\n')

arr =  [list(map(int,row.split())) for row in raw_arr]
arr.sort(key = lambda x: x[1])

```

```python

miny = -1
for y, x in arr:
    miny = max(miny, y)
# print(miny)


def calc(size):
    
    result = 0
    covered_x = 0

    for _, x in arr:
        if covered_x < x:
            covered_x = x + size - 1
            result +=1

    return result

start = miny
end = 10

ans = miny
while start <= end:
    mid = (start + end) // 2
    value = calc(mid)
    if value <= P:
        ans = mid
        end = mid - 1
    
    else:
        start = mid + 1

print(ans)
```
