```python
dp = [ 0 for _ in range(N)]
visited = [False for _ in range(N)]

def dfs(cur_idx):
    ans = len(arr[cur_idx])
    for next_idx in range(N):
        if visited[next_idx]:
            continue
        if arr[cur_idx][-1] == arr[next_idx][0]:
            visited[next_idx] = True
            ans = max(ans, ans +dfs(next_idx) )
            visited[next_idx] = False
    dp[cur_idx] = max(ans, cur_idx)
    return dp[cur_idx]

for n in range(N):
    if not dp[n]:
        visited[n] = True
        dfs(n)
        visited[n] = False
```
