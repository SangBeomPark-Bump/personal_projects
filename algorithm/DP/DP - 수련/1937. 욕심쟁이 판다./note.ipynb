```python
N = 4

raw_arr = """14 9 12 10
1 11 5 4
7 15 2 13
6 3 16 8""".split('\n')

arr =  [list(map(int,row.split())) for row in raw_arr]
```

```python
dp = [[0 for _ in range(N)]for _ in range(N)]

dx = (-1, 1, 0, 0)
dy = (0, 0, -1, 1)


#### dfs함수. 이동경로를 print해줌
def dfs(y, x):
    if dp[y][x]>0:
        return dp[y][x]
    dp[y][x] = 1
    for i in range(4):
        ### new_x, new_y
        ny = y + dy[i]
        nx = x + dx[i]
        ### 경계면이라면 for문 거르기
        if not(0 <= ny < N and 0 <= nx < N):
            continue
        if arr[ny][nx] > arr[y][x]:
            dp[y][x] = max(dp[y][x], dfs(ny, nx) + 1)
    return dp[y][x]


for i in range(N):
    for j in range(N):
        if not dp[i][j]:
            dfs(i, j)

ans = 0
for i in range(N):
    for j in range(N):
        ans = max(ans, dp[i][j])
print(ans)
```
