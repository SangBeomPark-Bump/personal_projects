```python
N = 3

raw_arr = """0 10 11
10 0 12
12 13 0""".split('\n')
arr =  [list(map(int,row.split())) for row in raw_arr]

string = 'YNN'

P = 3
```

```python
ybit = 0
yarr = []

temp = 1
for i in range(N):
    if string[i] == 'Y':
        ybit += temp
        yarr.append(i)
    temp *= 2

# print(ybit)
# print(yarr)
```

```python
def rankcalc(bit):
    value = 0
    while bit > 0:
        if bit %2 == 1:
            value +=1
        bit //= 2
    return value
```

```python
bitrank = [ [] for _ in range(N+1)]
bitrank

for bit in range(1<<N):
    bitrank[rankcalc(bit)].append(bit)

bitrank
```

```python
dp = [ [INF for _ in range(N)] for _ in range(1<<N)]


for i in yarr:
    for j in range(N):
        dp[ybit][j] = min(arr[i][j], dp[ybit][j])

import math
ans = INF
for rank in range(rankcalc(ybit)+1, N+1):
    for bit in bitrank[rank]:
        if bit & ybit != ybit:
            continue

        num = 0
        value = INF
        curarr = []

        valuebit = bit

        while valuebit>0:
            if valuebit%2 == 1:
                value = min(dp[bit - (1<<num)][num], value)
                curarr.append(num)
            valuebit //= 2
            num +=1

        if rank >= P:
            ans = min(ans, value)
            continue

        for i in curarr:
            for j in range(N):
                if (1<<j) & bit:
                    dp[bit][j] = 0
                    continue
                dp[bit][j] = min(value + arr[i][j], dp[bit][j])

print(ans)

```
