```python
N, M = 5, 7

raw_arr = """1 2 3
1 3 4
3 1 -7
2 3 2
3 4 1
4 2 -2
4 5 1""".split('\n')

arr =  [tuple(map(int,row.split())) for row in raw_arr]
```

```python
INF = int(1e9)
money_arr = [ -INF for _ in range(N+1) ]
money_arr[1] = 0
path = [0] * (N + 1)

for i in range(N):
    for start_loc, end_loc, money_gained in arr:
        cur_money = money_arr[start_loc] + money_gained

        if money_arr[end_loc] < cur_money :
            money_arr[end_loc] = cur_money
            path[end_loc] = start_loc
            if i == N-1:
                print(end_loc, start_loc)
                money_arr[end_loc] = INF
            
```
