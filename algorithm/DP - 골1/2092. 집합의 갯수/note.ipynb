```python
verbose = True
verboses = [True for _ in range(10)]
INF = float('inf')
MOD = int(1e6)
```

```python
T, A, S ,B =  list(map(int, "3 5 2 3".split()))
arr =  list(map(int, "1 2 2 1 3".split()))
```

```python
from collections import Counter


counter_dict = Counter(arr)
```

```python
dp = [[0 for _ in range(A+1)] for _ in range(len(counter_dict) +1)]
dp[0][0] = 1

for index, (c, freq) in enumerate(counter_dict.items(), start=1): ## a
    for a in range(A+1): # A번
        for i in range(freq+1): ## a와 같이 T번
            if a+i < A+1:
                dp[index][a+i] += dp[index-1][a]


print(sum(dp[-1][S:B+1]))
```

```python
dp = [[0 for _ in range(B+1)] for _ in range(len(counter_dict) +1)]
dp[0][0] = 1

for index, (c, freq) in enumerate(counter_dict.items(), start=1): ## a
    for a in range(A+1): # A번
        for i in range(freq+1): ## a와 같이 T번
            if a+i < B+1:
                dp[index][a+i] += dp[index-1][a]
                dp[index][a+i] %= MOD

print(sum(dp[-1][S:B+1]) % MOD)
```
