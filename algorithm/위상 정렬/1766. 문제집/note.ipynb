```python
start_degree_arr = [[[], 0] for _ in range(N+1)] 
for start, end in arr:
    start_degree_arr[start][0].append(end)
    start_degree_arr[end][1] +=1
```

```python
from collections import deque

ques = [deque()]

for idx, (_, degree) in enumerate(start_degree_arr[1:], start=1):
    if degree == 0:
        ques[0].append(idx)
```
