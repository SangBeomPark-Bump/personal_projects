```python
N, M = 10, 10

raw_arr = """....x.....
..........
..........
..x.......
..........
x...x.x...
.........x
...x......
........x.
.x...x....""".split('\n')

arr =  [list(number) for number in raw_arr]

# arr =  [list(map(int,row.split())) for row in raw_arr]
```

```python
for i in arr:
    print(i)
```

```python
dp = [ [-INF for _ in range(1 << M)] for _ in range(N+1)]
dp[0][0] = 0
```

```python
pos_bit = []
for bit in range(1<<10):
    cnt = 0
    testbit = bit
    while testbit:
        if testbit %2 == 1:
            cnt += 1
        else:
            cnt = 0
        if cnt == 2:
            break
        testbit //= 2
    if cnt == 2:
        continue
    else:
        pos_bit.append(bit)
# pos_bit
```

```python
impos_bitarr = [ 0 for _ in range(1<<10)]

for bit in pos_bit:
    impossible_set = set()
    bitindex = bit

    index = 0
    while bit:
        if bit %2:
            if index + 1 < M:
                impossible_set.add(index+1)
            if index -1 >=0:
                impossible_set.add(index-1)
        index +=1
        bit //= 2

    impos_bit = 0
    for impossbile_index in impossible_set:
        impos_bit |= 1<<impossbile_index
    impos_bitarr[bitindex] = impos_bit
# impos_bitarr
```

```python
bitrank = [0 for _ in range(1<<10)]
for bit in pos_bit:
    indexbit = bit
    cnt = 0
    while bit:
        if bit %2:
            cnt +=1
        bit //=2
    bitrank[indexbit] = cnt
# bitrank
```

```python
dp = [ [-INF for _ in range(1 << M)] for _ in range(N+1)]
dp[0][0] = 0

for n in range(1, N+1):
    xloc = 0
    for m in range(M):
        if arr[n-1][m] == 'x':
            xloc |= (1<<m)
    for cur_bit in pos_bit:
        if cur_bit >= 1<<M:
            break

        if dp[n-1][cur_bit] == -INF :
            continue
        impos_bit = impos_bitarr[cur_bit]
        impos_bit |= xloc

        for next_bit in pos_bit:
            if next_bit >= 1<<M:
                break

            if impos_bit & next_bit:
                continue
            # print(next_bit)
            dp[n][next_bit] = max(dp[n][next_bit], dp[n-1][cur_bit] + bitrank[next_bit])
    # print(dp[n])
print(max(dp[-1]))
```
