### BFS야 들어가라얍

```python
from collections import deque
def bfs(y, x):
    ### 여기는 앞에와 같음
    visited = [[0] * M for _ in range(N)]
    dy = (-1, 1, 0, 0)
    dx = (0, 0, -1, 1)
    ## 초기설정
    visited[y][x] = 1
    que = deque()
    ## y, x에 넣어준다
    que.append((y, x))

    ans = -1

    while que:
        y, x = que.popleft()
        factor = arr[y][x]
        for i in range(4):
            ny = y + dy[i] * factor
            nx = x + dx[i] * factor
            if not(0 <= ny < N and 0 <= nx < M):
                continue
            if arr[ny][nx] == 0:
                continue
            if not visited[ny][nx]:
                visited[ny][nx] = visited[y][x] +1
                que.append((ny, nx))
            else:
                return ans, None if not verbose else visited
    for row in visited:
        for num in row:
            ans = max(num, ans)
    return ans, None if not verbose else visited
print(bfs(0,0))
```

```python
yes, visitied =  bfs(0,0)
print(yes)
for i in visitied:
    print(i)
```

### DFS야 나와라얍

```python
### 방문 했는지 안했는지 알려줌
visited = [[False] * M for _ in range(N)]
dp = [[0] * M for _ in range(N)]

dy = (-1, 1, 0, 0)
dx = (0, 0, -1, 1)

#### dfs함수. 이동경로를 print해줌
def dfs(y, x):
    ### 일단 0번 print
    if dp[y][x]:
        return dp[y][x]
    
    if visited[y][x]:
        return float('INF')

    visited[y][x] = True
    factor = arr[y][x]
    max_num = 1
    for i in range(4):
        ### new_x, new_y
        ny = y + dy[i] * factor
        nx = x + dx[i] * factor
        ### 경계면이라면 for문 거르기
        if not(0 <= ny < N and 0 <= nx < M):
            continue
        ### 벽이라면 거르기
        if arr[ny][nx] == 0:
            continue
        ### 방문 안했다면!
        dp[ny][nx] = dfs(ny, nx)+1
        max_num = max(dp[ny][nx], max_num)
            
    visited[y][x] = False
    return max_num

ans = dfs(0,0)
if ans == float('inf'):
    print(-1)
else:
    print(ans)
```
