```python
N = 8


raw_arr = """1 8
3 9
2 2
4 1
6 4
10 10
9 7
7 6""".split('\n')


arr =  [tuple(map(int,row.split())) for row in raw_arr]
```

```python
arr
```

### 
1. 전깃줄끼리  디그리노드 쓰고
2. 위상정렬 고고씽

```python
arr.sort(key=lambda x : x[0])
arr
```

```python
dp = [ [] for _ in range(N) ]
degree = [ 0 for _ in range(N)]


for n in range(N):
    _, se = arr[n]
    for i in  range(n+1, N):
        _, fe = arr[i]
        if se > fe:
            dp[n].append(i)
            dp[i].append(n)
            degree[n] +=1
            degree[i] +=1

print(dp)
print(degree)
```

```python
value = 0
while True:
    cur_degree = max(degree)
    if cur_degree == 0:
        break
    value +=1
    cur_index = degree.index(max(degree))

    for next_index in dp[cur_index]:
        degree[next_index] -=1
        dp[next_index].remove(cur_index)

    dp[cur_index] = []
    degree[cur_index] = 0

print(value)
```

### dp식 풀이법

```python
is_crossed = [ [False for _ in range(N)] for _ in range(N)]

for n in range(N):
    _, se = arr[n]
    for i in  range(n+1, N):
        _, fe = arr[i]
        if se > fe:
            is_crossed[n][i] = True
            is_crossed[i][n] = True
```

```python
dp = [ [] for _ in range(N) ]
dp[0] = [set([0])]

from copy import deepcopy
n = 1

for n in range(1, N):
    dp[n] = deepcopy(dp[n-1])

    for i in range(len(dp[n-1])):
        append = True
        for j in dp[n-1][i]:
            if is_crossed[n][j]:
                append = False
        if append:
            dp[n][i].add(n)

    dp[n].append(set({n}))
```
