```python
prob = "acxcba"
N = len(prob)

def is_pelindrom(string):
    len_str = len(string)
    a = string[:len_str//2]
    if len_str %2 ==0:
        b = string[len_str//2:]
    else:
        b = string[(len_str//2)+1:]
    return a == b[::-1]
```

```python
## 삽입
## 삭제
## 교환
## 지들끼리교환
```

```python
dp = [ [] for _ in range(N+1)]

dp[0].append(['', 0])

def insert(string, index, num ):
        if is_pelindrom(string):
            dp[index+1].append([string, num])

for i in range(N):
    letter = prob[i]

    for prev_pel, num in dp[i]:
        cand = prev_pel + letter
        insert(cand, i, num )
        #삽입
        if prev_pel:
            temp = cand + prev_pel[0]
            insert(temp, i, num+1 )
            ## 교환
            temp = cand[:-1] + prev_pel[0]
            insert(temp, i, num+1)
        temp = letter + cand
        insert(temp, i, num+1)

        #삭제
        temp = cand[:-1]
        insert(temp, i, num+1)

        temp = cand[1:]
        insert(temp, i, num+1)

        #교환
        temp = letter + cand[1:]
        insert(temp, i, num+1)
```

```python
from collections import deque
MAX = float('inf')
di = [0, 1, 1]
dj = [-1, 0, -1]

string = prob

def cal_dp(target) :
  result = MAX
  dp = [[MAX]*len(string) for _ in range(len(string))]
  dp[0][-1] = 0
  q = deque([(0, len(string)-1, 0)])

  while q :
    i, j, dist = q.popleft()
    if i >= j :
      result = min(result, dist)
      continue
    if target[i] == target[j] :
      if dp[i+1][j-1] > dist :
        dp[i+1][j-1] = dist
        q.append((i+1, j-1, dist))
      continue
    for k in range(3) :
      ai, aj = i + di[k], j + dj[k]
      if dp[ai][aj] > dist + 1 :
        dp[ai][aj] = dist + 1
        q.append((ai, aj, dist+1))

  return result

answer = cal_dp(string)
for i in range(len(string)-1) :
  for j in range(i+1, len(string)) :
    if string[i] == string[j] :
      continue
    string[i], string[j] = string[j], string[i]
    answer = min(answer, cal_dp(string)+1)
    string[i], string[j] = string[j], string[i]

print(answer)
```
