```python
ans = 0
for y in range(N):
    for x in range(M):
        if arr[y][x] == 0:
            continue
        value = float('inf')
        if y == 0 or x == 0:
            value = 0
        else: 
            value = min(value, arr[y-1][x], arr[y][x-1], arr[y-1][x-1])
        arr[y][x] = value +1
        ans = max(value +1, ans)

# print(ans **2 )
```
