```python
gold = []


for i in range(2):
    a = ['4' for _ in range(i+1)]
    for bit in range(1 << (i+1)):
        index = i
        while bit:
            if bit %2:
                a[index] = '7'
            else:
                a[index] = '4'
            bit//=2
            index -=1
        b = int(''.join(a))
        gold.append(int(b))
```

```python
k = [0 for _ in range(X+1)]
index = 0
for g in range(X+1):
    if index < len(gold):
        if g == gold[index]:
            index +=1
    k[g] = index
k
```

```python
X = 100
dp = [[] for _ in range(X+1)]
# dp[0] = 0
cnt = 0
def dfs(x):
    global cnt
    cnt +=1
    if cnt == 100:
        return
    print(x)

    if dp[x]:
        return dp[x]

    if x in [1, 2, 3]:
        return []

    value = [0 for _ in range(int(1e5))]
    for i in range(k[x]-1, -1, -1):
        temp = dfs(x%gold[i])
        if temp:
            value =min(value, temp + [gold[i] for _ in  range(x//gold[i])], key=len)
            dp[x] = value

    return dp[x]

dfs(4)
```
