```python
N = 7
K = 3
```

```python
from collections import deque

deq = deque(range(1, N+1))
ans = []
length = 0
while deq:
    length +=1
    x = deq.popleft()

    if length != K:
        deq.append(x)
    else:
        ans.append(x)
        length = 0
```
