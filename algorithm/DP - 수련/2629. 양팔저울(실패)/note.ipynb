### 재도전

```python
available = [False for _ in range(40001)]

dp = [ [set() for _ in range(N)] for _ in range(N)]

for n in range(N):
    dp[n][n].add(ns[n])
    available[ns[n]] = True
```

```python
def setmaker(list1, list2):
    global available
    myset = set()
    for i in list1:
        for j in list2:
            if i < j:
                myset.add(j - i)
                available[j-i] = True
            elif i> j :
                myset.add(i - j)
                available[i - j] = True
            myset.add(i + j)
            available[i+j] = True
    return myset
```

```python
for factor in range(1, N+1):
    for start in range(N-factor):
        end = start  + factor
        
        for startfactor in range(factor):
            endfactor = factor - startfactor - 1
            dp[start][end] = dp[start][end].union(setmaker(dp[start][start+ startfactor], dp[end - endfactor][end]))
```
