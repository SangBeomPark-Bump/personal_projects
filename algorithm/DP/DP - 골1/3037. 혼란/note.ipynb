```python
N = 4
C = 10
MOD = int(1e9) + 7
```

```python
dp = [ [0 for _ in range(C+1)] for _ in range(N+1)]
dp[1][0] = 1

for n in range(2, N+1): ## N번
    for start_number in range(1, n+1): ## N번
        for c in range(C+1): ## C번
            if not dp[n-1][c]:
                continue
            base = n - start_number
            nextc = base + c
            if nextc > C:
                continue
            dp[n][nextc] += dp[n-1][c]
            dp[n][nextc] %= MOD
    print(dp[n])


print(dp[-1][-1])

```
