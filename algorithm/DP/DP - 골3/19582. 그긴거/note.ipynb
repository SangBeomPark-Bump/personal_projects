```python
N = 4


raw_arr = """10 3
2 5
3 5
20 8""".split('\n')


arr =  [list(map(int,row.split())) for row in raw_arr]
```

```python
arr
```

```python
cs = [0]
temp = 0

first_high = 0
max_high = -INF
not_first_max_high = -INF

for i in range(1,N):
    cer = arr[i][0]
    value = arr[i-1][1]
    temp += value
    cs.append(temp - cer)
    if temp - cer > 0:
        if not first_high:
            first_high = i
        else:
            not_first_max_high = max(not_first_max_high, temp - cer)
        max_high = max(max_high, temp - cer)

cs
```

```python
ans = first_high == 0
for i in range(first_high):
    if arr[i][1] >= max_high:
        ans = True
        break
if arr[first_high][1] >= not_first_max_high and not ans:
    ans = True

print("Kkeo-eok" if ans else "Zzz")
```
