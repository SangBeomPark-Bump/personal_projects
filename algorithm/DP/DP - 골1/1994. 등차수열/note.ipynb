```python
dp = [ [-1 for _ in range(N)] for _ in range(N)]
def recursive(start, end):
    if dp[start][end] >=0:
        return dp[start][end]
    dp[start][end] = 0
    for next_end in range(end+1, N):
        if arr[next_end] - arr[end] == arr[end] - arr[start]:
            dp[start][end] = 1 + recursive(end, next_end)
    return dp[start][end]

max_len = -1
for i in range(N-1):
    for j in range(i+1,N):
        max_len = max(max_len, recursive(i,j))
print(max_len +2)
```
