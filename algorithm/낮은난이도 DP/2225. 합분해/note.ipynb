```python
dp = [[0 for _ in range(K+1)] for _ in range(N+1) ]
dp[0][0] = 1
```

```python
for n in range(N+1):
    for k in range(1,K+1):
        for i in range(n+1):
            dp[n][k] += dp[i][k-1]
```
