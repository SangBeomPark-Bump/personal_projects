```python

INF = float('inf')
```

```python
N = 4
INF = N * N + 1
raw_arr = """4 3 2 1
0 0 0 0
0 0 9 0
1 2 3 4""".split('\n')

arr =  [list(map(int,row.split())) for row in raw_arr]
dydx = ( (1, 0), (-1, 0), (0, 1), (0, -1))
```

```python
from collections import deque
for y in range(N):
    for x in range(N):
        if arr[y][x] == 9:
            cur_loc = (y, x)
            arr[y][x] = 0


def eat(loc, size):
    cnt = 0
    visited = [[False for _ in range(N)] for _ in range(N)]
    time = 0

    que = deque()
    que.append((loc, 0))
    visited[loc[0]][loc[1]] = True

    result = (INF, INF, INF)

    while que:
        cnt +=1
        if cnt >= 100:
            return
        (y, x), cur_time = que.popleft()
        if cur_time < time:
            break

        if arr[y][x]:
            time = cur_time
            result = min(result, (time, y, x))
            continue

        for dy, dx in dydx:
            ny = y + dy
            nx = x + dx

            if not( 0 <= ny < N and 0 <= nx < N):
                continue
            if visited[ny][nx]:
                continue
            if arr[ny][nx] > size:
                continue
            visited[ny][nx] = True
            que.append(((ny, nx), cur_time + 1))

    return result


spent_time = 0
eaten_fish = 0
cur_size = 2

while True:
    cur_time, y, x = eat(cur_loc, cur_size)
    print(cur_time, y, x)
    if cur_time == INF:
        break

    arr[y][x] = 0
    cur_loc = (y, x)

    spent_time += cur_time
    eaten_fish += 1

    if eaten_fish == cur_size:
        eaten_fish = 0
        cur_size +=1
```
