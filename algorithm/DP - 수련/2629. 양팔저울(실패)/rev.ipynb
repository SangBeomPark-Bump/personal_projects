```python
N = 2
ns = [0] + list(map(int, "1 4".split()))


M = 2
ms = list(map(int, "3 2".split()))
```

```python
dp = [[False for _ in range(15001)] for _ in range(N+1)]
dp[0][0] = True
```

```python
for n in range(1, N+1):
    for d in range(15001):
        if dp[n-1][d]:
            dp[n][d] = True 
            dp[n][d+ns[n]] = True
            dp[n][abs(d-ns[n])] = True

print(*["Y" if dp[-1][m] else "N"  for m in ms])
```
