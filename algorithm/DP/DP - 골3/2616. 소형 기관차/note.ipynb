```python
N = 7
arr = [0] + list(map(int, "35 40 50 10 30 45 60".split()))
M = 2
```

```python
dp = [ [0,0,0] for _ in range(N+1)]

cur_sum = 0
for n in range(1, N+1):
    cur_sum += arr[n]
    if n >= M:
        cur_sum -= arr[n - M]
    if M<= n <= N-2*M:
        dp[n][0] = max(dp[n-1][0] , cur_sum)
    if 2*M <= n <= N-M:
        dp[n][1] = max(dp[n-1][1], cur_sum + dp[n-M][0])
    if n >= 3 * M:
        dp[n][2] = max(dp[n-1][2], cur_sum + dp[n-M][1])

for i in dp:
    print(i)
```

```python
dp = [ [0 for _ in range(4)] for _ in range(N+1)]

cur_sum = 0
for n in range(1, N+1):
    cur_sum += arr[n]
    if n >= M:
        cur_sum -= arr[n - M]
    for i in range(1, 4):
        dp[n][i] = max(dp[n-1][i] , cur_sum + dp[n-M][i-1])

for i in dp:
    print(i)
```
