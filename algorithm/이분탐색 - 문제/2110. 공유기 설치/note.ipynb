```python
N, C = 5, 3


raw_arr = """1
2
8
4
9""".split('\n')

arr =  [int(number) for number in raw_arr]
arr.sort()
```

```python
arr
```

```python
ans = 0
start = 0
end = arr[-1] - arr[0]

while start <= end:
    
    mid = (start + end) // 2
    
    curhouse = arr[0]

    c = 1
    for n in range(1, N):
        if arr[n] - curhouse >= mid:
            curhouse = arr[n]
            c += 1
    
    if c < C:
        end = mid - 1
    else:
        ans = max(ans, mid)
        start = mid + 1

print(ans)
```
