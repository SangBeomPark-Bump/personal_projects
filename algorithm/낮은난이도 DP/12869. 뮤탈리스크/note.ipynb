```python
from collections import deque
from itertools import permutations



dp = [ [ [ 0 for _ in range(arr[2]+1) ] for _ in range(arr[1]+1) ] for _ in range(arr[0]+1)]
# dp[arr[0]][arr[1]][arr[2]] = 0

def dfs(n1, n2, n3):

    if n1 == 0 and n2 == 0 and n3 == 0:
        dp[0][0][0] = 0
        return 0

    my_list = list(permutations([9, 3, 1], 3))

    for d1, d2, d3 in my_list:
        next_num1 = n1 - d1 if n2 > d1 else 0
        next_num2 = n2 - d2 if n2 > d2 else 0
        next_num3 = n3 - d3 if n3 > d3 else 0

        if dp[next_num1][next_num2][next_num3]:
            continue
        dp[n1][n2][n3] = dfs(n1, n2, n3) + 1
        return dp[n1][n2][n3]
```
