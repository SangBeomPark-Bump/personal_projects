```python
from collections import deque
que = deque()

time_arr2 = [0] * (N + 1)

for b_name, b_time in enumerate(end_arr[1:], start=1):
    if b_time == 0:
        que.append(b_name)
        time_arr2[b_name] = time_arr[b_name]

while que:
    cur_building = que.popleft()
    if verbose:
        print(cur_building)

    for next_building in start_arr[cur_building]:
        time_arr2[next_building] = max(time_arr2[next_building],time_arr2[cur_building] + time_arr[next_building])
        end_arr[next_building] -=1
        if end_arr[next_building] == 0:
            que.append(next_building)
            if verbose:
                print(que, cur_building, next_building, time_arr2[next_building])

    if cur_building == W:
        break

    print(time_arr2[W])
```

```python
from collections import deque

T = int(input())

for _ in range(T):
    N, K = map(int, input().split())
    D = [0] + list(map(int, input().split()))## timearr

    arr = [[] for _ in range(N + 1)]
    degree = [0] * (N + 1)
    for _ in range(K):
        X, Y = map(int, input().split())
        arr[X].append(Y)
        degree[Y] += 1

    W = int(input())

    que = deque()
    time = [0] * (N + 1)

    for i in range(1, N + 1):
        if degree[i] == 0:
            que.append(i)
            time[i] = D[i]

    while que:
        idx = que.popleft()

        for a in arr[idx]:
            degree[a] -= 1
            time[a] = max(time[idx] + D[a], time[a])
            if degree[a] == 0:
                que.append(a)

        if idx == W:
            break

    print(time[W])
```
