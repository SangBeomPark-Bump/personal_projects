```python
import math
P = int(5e5)
rp = math.floor(math.sqrt(P))

is_prime = set( range(2, P+1))

for n in range(2, rp+1):
    if n in is_prime:
        for j in range(n * 2,P+1,n):
            if j in is_prime:
                is_prime.remove(j)
```

```python
dictdp = defaultdict(int)
dictdp[0] = 1

for n in range(1, len(arr2)+1):
    nextdictdp = defaultdict(int)
    number, many = arr2[n-1]
    for prev_number in dictdp.keys():
        nextdictdp[prev_number] += dictdp[prev_number]
        for k in range(1, many+1):
            cur_number = number * k
            nextdictdp[prev_number + cur_number] += dictdp[prev_number]
    dictdp = nextdictdp
```
