```python
N = 250111
```

```python
dp = [INF for _ in range(100)]
dp[0] = 0

for n in range(1, 100):
    case1 = dp[n-1]
    case2 = INF
    case3 = INF
    if n>=10:
        case2 = dp[n-10]
    if n >= 25:
        case3 = dp[n-25]
    
    dp[n] = min(case1, case2, case3) +1
```

```python
x= 0
while N:
    x += dp[N %100]
    N //= 100
print(x)
```
