```python
N = 11
arr =  list(map(int, "1 2 6 7 11 12 13 15 17 20 23".split()))
```

```python
# X = int(1e6)
X = 30

great = [ False for _ in range(X+1)]
value_to_index = dict()

for index, value in enumerate(arr):
    great[value] = True
    value_to_index[value] = index
# dp

dp = [[0 for _ in range(N)] for _ in range(N)]
dp
```

```python
from itertools import combinations

max_score = 0
for first, second in combinations(range(N), 2):
    d = arr[second] - arr[first] 
    third = arr[second] + d
    if third > X:
        continue
    if not dp[third]:
        continue
    # print(first, second, third)
    cur_score = arr[first] + arr[second] + third
    next_number = third + d
    # print(next_number)
    while True:
        if next_number > X:
            break
        if not dp[next_number]:
            break
        cur_score += next_number
        next_number += d
    # print(arr[first], arr[second], cur_score)
    max_score = max(cur_score, max_score)
print(max_score)

```

```python
# X = int(1e6)
X = 30
value_to_index = dict()

for index, value in enumerate(arr):
    value_to_index[value] = index

dp = [[0 for _ in range(N)] for _ in range(N)]
```

```python
max_score = 0
for first in range(N-1):
    for second in range(first+1, N):
        if dp[first][second]:
            continue
        d = arr[second] - arr[first]
        dp[first][second] = arr[first] + arr[second]
        third_number = arr[second] + d
        if third_number > X:
            continue
        if third_number not in value_to_index.keys():
            continue

        cur_score = arr[first] + arr[second] + third_number
        start, end = second, value_to_index[third_number]
        dp[start][end] = cur_score
        next_number = third_number + d

        while True:
            if next_number > X:
                break
            if next_number not in value_to_index.keys():
                break
            cur_score += next_number
            start, end = end, value_to_index[next_number]
            dp[start][end] = cur_score
            next_number += d
        # print(arr[first], arr[second], cur_score)
        max_score = max(cur_score, max_score)
    print(max_score)

```
