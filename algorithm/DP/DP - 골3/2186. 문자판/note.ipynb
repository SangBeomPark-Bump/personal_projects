```python
dp = [[ [ 0 for _ in range(end) ] for _ in range(M)] for _ in range(N)]


def recursive_true(y, x, depth):
    if depth == end-1:
        dp[y][x][depth]= 1
        return 1
    
    if dp [y][x][depth]:
        return dp[y][x][depth]

    for dy, dx in temp:
        ny = y + dy
        nx = x + dx
        if ny >= N or ny < 0 or nx >= M or nx<0:
            continue

        if arr[ny][nx] == ans[depth+1]:
            dp[y][x][depth] += recursive_true(ny, nx, depth+1)
    
    return dp[y][x][depth]

value = 0
for y in range(N):
    for x in range(M):
        if arr[y][x] == ans[0]:
            value += recursive_true(y, x, 0)
print(value)

```
