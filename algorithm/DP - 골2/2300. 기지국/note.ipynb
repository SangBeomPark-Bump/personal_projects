```python
N = 7

raw_arr = """-2 -1
-4 -1
0 2
3 1
5 1
1 -1
8 -1""".split('\n')


arr =  [list(map(int,row.split())) for row in raw_arr]
arr.sort(key= lambda x: x[0])
```

```python
dp = [ INF for _ in range(N) ]
```

```python
dp = [ INF for _ in range(N) ]
def recursive(start):
    if start >= N:
        return 0

    if dp[start] != INF:
        return dp[start]

    x, _ = arr[start]
    maxx = 0
    maxy = 0

    for i in range(start, N):
        nextx, nexty = arr[i]
        maxx = max(maxx,abs((nextx - x)) )
        maxy = max(maxy, abs(nexty) * 2)

        dp[start] = min( dp[start], max(maxx, maxy) + recursive(i+1))
    return dp[start]


print(recursive(0))


```
