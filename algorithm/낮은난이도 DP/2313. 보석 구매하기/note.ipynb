```python
sarr = [0]

temp = 0
for i in range(N):
    temp += arr[i]
    sarr.append(temp)
```

```python
dp = [ 0 for _ in range(N+1)]

last_minus = 0
temp = 0

start = 0
end = 0

for n in range(1,N+1):
    case1 = dp[n-1] + arr[n]
    case2 = dp[n-1]

    temp += arr[n]
    case3 = temp

    if case1 >= case3 and case1 >= case2:
        end = n
        dp[n] = case1
    
    elif case3 >= case1 and case3 >= case2 and last_minus:
        end = n
        start = last_minus
        last_minus = 0
        temp = 0
        dp[n] = case3

    elif case2 >= case1 and case2 >= case3:
        dp[n] = case2

    if arr[n]<0:
        last_minus = n
        temp = 0
    print(case1, case2, case3, dp[n])
```

```python
arr =  list(map(int, "1 2 3 -6 6".split()))
N = len(arr)
```

```python
INF = float("inf")
start = 0
end = 0
value = 0

max_value = -INF
max_start = 0
max_end = 0

while end<N:
    value += arr[end]

    if value> max_value or (value == max_value and max_end - max_start > end-start):
        max_value = value
        max_start = start
        max_end = end

    if value <=0:
        start = end+1
        value = 0
    end+=1
print(max_value, max_start+1, max_end +1)
```
