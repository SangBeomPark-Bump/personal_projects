```python
dp = [ [0 for _ in range(P+1)] for _ in range(N+1)]

dp[N][M] = perm(N, M)


for x in range(N, -1, -1):
    for p in range(1, P+1):
        if x > p:
            dp[x][p] = 0
            continue
        if x > 0:
            dp[x][p] += x * dp[x-1][p-1]
        if x<M and (N - min(p-1, M) - x) >0 :
            dp[x][p] += (N - min(p-1, M) - x) * dp[x][p-1]

for i in dp:
    print(i)
```
