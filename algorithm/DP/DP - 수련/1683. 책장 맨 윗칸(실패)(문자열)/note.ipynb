```python
N = 12


raw_arr = """117 The Pearl
42 L'Etranger
135 Hamlet
100 War And Peace
55 The Prince
175 Of Mice And Men
200 The Grapes Of Wrath
120 Ulysses
87 For Whom The Bell Tolls
10 Oliver Twist
50 Snow Crash
110 Great Expectations""".split('\n')



arr =  [ [int(row.split()[0]), ''.join(row.split()[1:])] for row in raw_arr]
```

```python
def extract_english_letters(text):
    return ''.join(char for char in text if char.isalpha() and char.isascii()).lower()

super_arr = list(map(lambda x: [x[0] ,extract_english_letters(x[1])], arr))
super_arr
```

```python
scores = [ row[0] for row in arr]
books = [ row[1] for row in arr]

adjustied_books = list(map(extract_english_letters, books))
```

```python
dp_books = [ [[] for _ in range(N)] for _ in range(11)]
dp_books[1] = [ [num] for num in range(N) ]
dp_books[1]
```

```python
dp_scores = [ [0 for _ in range(N)] for _ in range(11)]
dp_scores[1] = [ score for score in scores]
dp_scores
```

```python
def is_possible(book1, book2):
    if len(book1) < len(book2):
        temp = book1
    else:
        temp = book2
    for i in range(len(temp)):
        if book1[i] == book2[i]:
            return False
    return True
```

```python


for d in range(2, 10):
    for prev_books_nums in dp_books[d-1]: ## 2500
        prev_books = []
        prev_score = 0

        for prev_book_num in prev_books_nums: ## 10
            prev_books.append(adjustied_books[prev_book_num])
            prev_score = dp_scores[d-1][prev_book_num]
        
            for next_book_num in range(N): # 2500
                next_book = adjustied_books[next_book_num]
                next_score = scores[next_book_num]

                if next_book not in (prev_books): # 10
                    cand = prev_books[::1]
                    cand.append(next_book)
                    cand.sort()
                    next_book_index = cand.index(next_book)

                    if prev_score + next_score > dp_scores[d][prev_book_num]:
                        bool1 = True
                        bool2 = True
                        if next_book_index>0:
                            if not is_possible(cand[next_book_index-1], next_book) :
                                bool1 = False
                        if next_book_index<len(cand)-1 :
                            if not  is_possible(cand[next_book_index+1], next_book) :
                                bool2 = False
                        # print(cand[next_book_index-1], next_book, bool1, bool2, sep='\n')
                        # print('-' * 50)
                        if bool1 and bool2:
                            dp_scores[d][prev_book_num] = prev_score + next_score
                            dp_books[d][prev_book_num] = prev_books_nums + [next_book_num]
    print(dp_scores[d])
```
