```python
U = K**N

H = N // 5
M = N % 5
u = U

c = 1
X = (2 ** H) *(M+1) * (K ** M)
pi = K ** 0


for h in range(1, H+1):
    u -= X * c * pi

    pi * (K ** h)
    c *= (H-h+1)
    c //= h
u
```
