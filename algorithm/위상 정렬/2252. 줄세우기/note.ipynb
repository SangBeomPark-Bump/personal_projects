```python
N , M = 4, 2


raw_arr = """4 2
3 1""".split('\n')

arr =  [tuple(map(int,row.split())) for row in raw_arr]
```

```python
ps = [[number, [], 0] for number in range(N+1)]
# ps_end = [[] for _ in range(N+1)]


for start, end in arr:
    ps[start][1].append(end)
    ps[end][2] +=1

ps

```

```python
from collections import deque

que = deque()

for i in ps[1:]:
    if i[2] == 0:
        que.append(i)

que

result = []
while que:
    cur_num, ends, _ =  que.popleft()
    result.append(cur_num)
    for end_num in ends:
        ps[end_num][2] -=1
        if ps[end_num][2] == 0:
            que.append(ps[end_num])
print(' '.join(map(str, result)))
```
