```python
N = 3


raw_arr = """1 2 3
4 5 6
4 9 0""".split('\n')



arr =  [list(map(int,row.split())) for row in raw_arr]
```

```python
dp = [ [[0,0] for _ in range(3)] for _ in range(N)]
dp[0] = [ [i,i] for i in arr[0]]
# for i in dp:
#     print(i)

```

```python
n = 1
for i in range(3):
    tempmax = dp[n-1][1][0]
    tempmin = dp[n-1][1][1]

    if i >0:
        tempmax = max(dp[n-1][2][0], tempmax)
        tempmin = min(dp[n-1][2][1], tempmin)

    if i<2:
        tempmax = max(tempmax, dp[n-1][0][0])
        tempmin = min(tempmin, dp[n-1][0][1])

    dp[n][i][0] =  arr[n][i] + tempmax
    dp[n][i][1] = arr[n][i] + tempmin
print(dp[n])
```

### 메모리 절약정신

```python
dp = [ [i,i] for i in arr[0]]


for n in range(1, N):
    [(amax, amin), (bmax, bmin), (cmax, cmin)] = dp
    for i in range(3):
        tempmax = bmax
        tempmin = bmin
        if i >0:
            tempmax = max(cmax, tempmax)
            tempmin = min(cmin, tempmin)

        if i<2:
            tempmax = max(tempmax, amax)
            tempmin = min(tempmin, amin)

        dp[i][0] =  arr[n][i] + tempmax
        dp[i][1] = arr[n][i] + tempmin
    
print(max(dp, key= lambda x : x[0])[0],min(dp, key= lambda x : x[1])[1] )

```

### 메모리 이슈

```python
dp = [ [i,i] for i in arr[0]]


for n in range(1, N):
    dp = [ 
        [ arr[n][0]+ max(dp[0][0], dp[1][0]) ,arr[n][0]+ min(dp[0][1], dp[1][1])], 
        [arr[n][1] + max(dp[0][0], dp[1][0], dp[2][0]), arr[n][1] + min(dp[0][1], dp[1][1], dp[2][1])], 
        [ arr[n][2]+ max(dp[2][0], dp[1][0]) ,arr[n][2]+ min(dp[2][1], dp[1][1])], 
    ]
    
print(max(dp, key= lambda x : x[0])[0],min(dp, key= lambda x : x[1])[1] )
```

```python
dpmax = arr[0][::]
dpmin = arr[0][::]


for n in range(1, N):
    amax, bmax, cmax =  dpmax
    amin, bmin, cmin =  dpmin
    for i in range(3):
        tempmax = bmax
        tempmin = bmin
        if i >0:
            tempmax = max(cmax, tempmax)
            tempmin = min(cmin, tempmin)

        if i<2:
            tempmax = max(tempmax, amax)
            tempmin = min(tempmin, amin)

        dpmax[i] = arr[n][i] + tempmax
        dpmin[i] = arr[n][i] + tempmin

print(max(dpmax), min(dpmin))
```
