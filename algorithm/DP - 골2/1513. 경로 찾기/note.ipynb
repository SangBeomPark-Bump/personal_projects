```python
N, M, C = 6, 4, 2

raw_arr = """5 3
3 2""".split('\n')


arr =  [list(map(int,row.split())) for row in raw_arr]
```

```python
degree = [ [0 for _ in range(M)] for _ in range(N)]

for c in range(C):
    cy, cx = arr[c]
    degree[cy-1][cx-1] = c+1
degree
```

```python
dp = [ [[[0 for _ in range(C+1)] for _ in range(C+1)] for _ in range(M)] for _ in range(N)]
dp[0][0][0][0] = 1
```

```python
dp = [ [[[0 for _ in range(C+1)] for _ in range(C+1)] for _ in range(M)] for _ in range(N)]
dp[0][0][0][0] = 1

move = ([1, 0], [0, 1])
for y in range(N):
    for x in range(M):
        for manyc in range(C+1):
            for maxc in range(C+1):
                for i in range(2):
                    dy, dx = move[i]
                    ny = y + dy
                    nx = x + dx
                    if ny <= N-1 and nx <= M-1:
                        if degree[ny][nx] and degree[ny][nx] < maxc:
                            continue
                        cur_maxc = 0
                        cur_maxc = max(degree[ny][nx], maxc)

                        cur_manyc = manyc
                        if degree[ny][nx]:
                            cur_manyc = manyc + 1
                        if cur_manyc <= C:
                            dp[ny][nx][cur_manyc][cur_maxc] += dp[y][x][manyc][maxc]

print(*map(sum, dp[-1][-1]))
```
