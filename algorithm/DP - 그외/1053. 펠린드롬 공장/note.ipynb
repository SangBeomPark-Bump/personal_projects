```python
prob = "acxcba"
N = len(prob)

def is_pelindrom(string):
    len_str = len(string)
    a = string[:len_str//2]
    if len_str %2 ==0:
        b = string[len_str//2:]
    else:
        b = string[(len_str//2)+1:]
    return a == b[::-1]
```

```python
## 삽입
## 삭제
## 교환
## 지들끼리교환
```

```python
dp = [ [] for _ in range(N+1)]

dp[0].append(['', 0])

def insert(string, index, num ):
        if is_pelindrom(string):
            dp[index+1].append([string, num])

for i in range(N):
    letter = prob[i]

    for prev_pel, num in dp[i]:
        cand = prev_pel + letter
        insert(cand, i, num )
        #삽입
        if prev_pel:
            temp = cand + prev_pel[0]
            insert(temp, i, num+1 )
            ## 교환
            temp = cand[:-1] + prev_pel[0]
            insert(temp, i, num+1)
        temp = letter + cand
        insert(temp, i, num+1)

        #삭제
        temp = cand[:-1]
        insert(temp, i, num+1)

        temp = cand[1:]
        insert(temp, i, num+1)

        #교환
        temp = letter + cand[1:]
        insert(temp, i, num+1)
```
