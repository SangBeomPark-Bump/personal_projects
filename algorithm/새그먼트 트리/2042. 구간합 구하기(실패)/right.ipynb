```python
N, M, K = 5, 2, 2
raw_arr_n = """1
2
3
4
5""".split('\n')

arr_n =  [int(number) for number in raw_arr_n]

raw_arr_o = """1 3 6
2 2 5
1 5 2
2 3 5""".split('\n')

arr_o =  [tuple(map(int,row.split())) for row in raw_arr_o]
```

```python

arr = arr_n

tree = [0] * (4 * N)

def init(node, start, end):
    # print(node, start, end)
    if start == end:
        tree[node] = arr[start]
        return

    mid = (start + end) // 2
    init(2 * node, start, mid)
    init(2 * node + 1, mid + 1, end)
    tree[node] = tree[2 * node] + tree[2 * node + 1]

init(1, 0, N-1)
```

```python
def update(node, start, end, idx, diff):
    if idx < start or end < idx:
        return

    tree[node] += diff

    if start != end:
        mid = (start + end) // 2
        update(2 * node, start, mid, idx, diff)
        update(2 * node + 1, mid + 1, end, idx, diff)
```

```python
def query(node, start, end, left, right):
    if right < start or end < left:
        return 0

    if left <= start and end <= right:
        return tree[node]

    mid = (start + end) // 2
    q1 = query(2 * node, start, mid, left, right)
    q2 = query(2 * node + 1, mid + 1, end, left, right)
    return q1 + q2 
```

```python
for a, b, c in arr_o:
    b -= 1
    if a == 1:
        diff = c - arr[b]
        arr[b] = c
        update(1, 0, N-1, b, diff)
    elif a == 2:
        c -= 1
        print(query(1, 0, N-1, b, c))
```
