```python
string = 'acattgagtc'
N = len(string)

```

```python
INF = float("INF")
dp = [ [0 for _ in range(N+1)] for _ in range(N+1)]
letter_dict = {"a" : "c", "g" : "t"}

def recursive(start, end):
    
    if start>=N:
        return 0

    if end == start:
        dp[start][end] = 0
        return 0

    if dp[start][end] != 0:
        return dp[start][end]

    letter = string[start]

    if letter == 'a' or  letter == 'g':
        for temp_end in range(end-1, start, -1):
            if string[temp_end] == letter_dict[letter]:
                left = recursive(start+1, temp_end)
                right = recursive(temp_end+1, end)
                # print(left, right, start, temp_end, end)
                dp[start][end] = max(left + 2 + right, dp[start][end])
    dp[start][end] = max(dp[start][end], recursive(start+1, end) + recursive(0, start+1))
    print(start, end, string[start:end], dp[start][end])
    return dp[start][end]

recursive(0,N)
```
