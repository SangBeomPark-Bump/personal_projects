```python
N, M = 11, 20

raw_arr = """7
4
2
3
2
5
1
12
7
5
6""".split('\n')

arr =  [0] + [int(number) for number in raw_arr]
```

```python
arr
```

```python
INF = float("INF")
dp = [ [INF for _ in range(M+1)] for _ in range (N+1)]
dp[0][M] = 0
# print(arr)
for n in range(1, N+1):
    cur_num = arr[n]
    for m in range(M+1):
        if m + 1 + cur_num <=M:
            dp[n][m+1+cur_num] = min(dp[n][m+1+cur_num], dp[n-1][m])
        dp[n][cur_num] = min(dp[n][cur_num], dp[n-1][m] + (M - m) **2)
    # print(dp[n])
print(min(dp[-1]))



```
