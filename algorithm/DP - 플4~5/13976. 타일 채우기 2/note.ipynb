```python
def matmul(A1, A2):
    ans = [ [0, 0], [0,0]]
    for i in range(2):
        for j in range(2):
            for k in range(2):
                ans[i][j] += A1[i][k] * A2[k][j]
            ans[i][j] %= MOD
    return ans
```

```python
def recursive(n):
    if n == 1:
        return [[4, -1], [1, 0]]
    if n == 0:
        return [[1, 0], [0, 0]]
    
    matrix = recursive(n//2)
    result = matmul(matrix, matrix)

    if n % 2:
        result = matmul(result, [[4, -1], [1, 0]])

    return result

print(sum(recursive(4)[0]) % MOD)
```
