```python
N = 5
K = 1
K -=1
```

```python

smallx = 0
samex = 0
bigx = 0
for x in range(N):
    if x < K:
        for y in range(K):
            x, y = sorted((x, y))
            smallx += fibonachi[x] * fibonachi[K - y -1] * fibonachi[N - K -1]
    elif x == K:
        samex += fibonachi[K] * fibonachi[N-K-1]
    else:
        for y in range(K+1, N):
            y, x = sorted((y, x))
            bigx += fibonachi[K] * fibonachi[y - K - 1] * fibonachi[N - x -1]
```

```python
print(bigx)
print(smallx)
print(samex)
```
