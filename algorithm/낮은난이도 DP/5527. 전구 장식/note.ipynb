```python
N = 5
arr = [-1] + list(map(int, "1 1 0 1 1".split()))
```

```python
dp = [0 for _ in range(N+1)]

for n in range(1, N+1):
    if arr[n] != arr[n-1]:
        dp[n] = dp[n-1] +1
    else:
        dp[n] = 1
dp
```

```python
value = 0
for n in range(1, N+1):
    value = max(value, dp[n] + dp[n - dp[n]] + dp[n-dp[n] - dp[n - dp[n]]])
print(value)
```
