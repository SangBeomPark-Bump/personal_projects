```python

def belman():
    INF = float("inf")
    dist = [INF] * (N+1)
    dist[1] = 0
    for i in range(1, N+1):
        for cur_loc, end_loc, spent_time in arr:
            if dist[cur_loc] == INF:
                continue
            next_dist = dist[cur_loc] + spent_time
            if dist[end_loc] > next_dist :
                dist[end_loc] = next_dist
                if i == N:
                    print("Yes")
                    return dist
    print("NO")
    return dist
```
