```python
V, E = 4, 5


raw_arr = """1 2 1
2 3 2
1 4 10
1 3 3
3 4 5""".split('\n')

arr =  [tuple(map(int,row.split())) for row in raw_arr]
arr.sort(key=lambda x : x[2])
```

```python
p = list(range(V+1))
p
```

```python
def find(a):
    if a != p[a]:
        p[a] = find(p[a])
    return p[a]

def union(a, b, value):
    pa = find(a)
    pb = find(b)

    if pa == pb:
        return 0
    if pa< pb:
        pb, pa = pa, pb

    p[pa] = pb
    return value
```

```python
p = list(range(V+1))
ans = 0
for start, end, value in arr:
    ans += union(start, end, value)
print(ans)
```
