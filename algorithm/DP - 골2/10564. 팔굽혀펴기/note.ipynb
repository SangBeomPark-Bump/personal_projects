```python
N, M = 6, 2
arr =  list(map(int, "3 1".split()))
```

```python
dp = [ -1 for _ in range(N+1)]
dp[0] = 0
for n in range(N+1):
    if dp[n] == -1:
        continue
    for m in range(M):
        newn = n + dp[n] + arr[m]
        if newn > N:
            continue
        dp[newn] = max(dp[newn], newn - n)

dp
```

```python
dp = [ [ False for _ in range(n+1)] for n in range(N+1)]
dp[0][0] = True
for n in range(N+1):
    for score in range(n+1):
        if not dp[n][score]:
            continue
        for m in range(M):
            newn = n + score + arr[m]
            if newn > N:
                continue
            dp[newn][newn-n] = True

ans = -1
for n in range(N+1):
    if dp[-1][n]:
        ans = max(ans, n)
print(ans)
```
