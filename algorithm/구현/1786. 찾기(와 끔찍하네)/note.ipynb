```python
def get_pi(p: str):
    m = len(p)
    j = 0
    pi = [0] * m

    for i in range(1, m):
        while j > 0 and p[i] != p[j]:
            j = pi[j - 1]

        if p[i] == p[j]:
            j += 1
            pi[i] = j

    return pi

PI = get_pi(P)
PI
```
