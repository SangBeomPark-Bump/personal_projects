```python
def query(left, right):
    left += size
    right += size 
    ans = 0
    while left <= right:
        if left % 2 == 1:
            ans += tree[left]
            left += 1
        left //= 2

        if right % 2 == 0:
            ans += tree[right]
            right -= 1
        right //= 2

    return ans
```
