```python
N = int(input())
dp = [[0 for _ in range(14)] for _ in range(N+1)]


dp[0][0] = 1

for i in range(N):
    for j in range(13):
        for k in range(4):
            if i + k <= N:
                dp[i+k][j+1] += dp[i][j] * combination_table[4][k]

idx = 1
ans = 0

while 4*idx <= N:
    temp = combination_table[13][idx]
    ans += temp * dp[N-4*idx][13-idx]
    ans %= div
    idx +=1

print(ans%div)
```
