```python
from heapq import heapify, heappush, heappop
mrr = list(map( lambda x: -x, arr))
heapify(mrr)

diff = INF
min_num = min(arr)

for _ in range(M):
    cur_max_num =  heappop(mrr) * -1
    diff = min(diff, cur_max_num - min_num)
    print(cur_max_num, min_num, diff, mrr)

    cur_max_num /=2
    min_num = min(cur_max_num, min_num)
    heappush(mrr, cur_max_num * -1)

```
