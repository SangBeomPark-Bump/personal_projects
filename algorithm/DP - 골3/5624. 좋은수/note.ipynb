```python
N = 6
arr =  list(map(int, "1 2 3 5 7 10".split()))
```

```python
from collections import deque



dp = [deque() for _ in range(4)]
dp
```

```python
ans = 0

dp = [deque() for _ in range(4)]

for n in range(N-1):
    a = arr[n]




    dp[0].append(arr[n] * 3)
    if n >=1:
        b = arr[n-1]
        dp[1].append(a * 2 + b)
        dp[2].append(a + b  * 2 )
    if n>=2:
        c = arr[n-2]
        dp[1].append(a * 2 + c)
        dp[2].append(a + c * 2)
        dp[3].append(a + b + c)

    if n>=3:
        for i in range(4):
            dp[i].popleft()
        for i in range(1, 3):
            dp[i].popleft()
    
    check = False
    for que in dp:
        for number in que:
            if number == arr[n+1]:
                check = True

    if check:
        ans +=1
    print(dp, arr[n+1])

print(ans)
```
