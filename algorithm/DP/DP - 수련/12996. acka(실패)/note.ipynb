```python

from collections import deque
dp = [[ [ [0 for _ in range(C+1)] for _ in range(B+1)] for _ in range(A+1)] for _ in range(S+1)]

que = deque()
dp[0][0][0][0] = 1
que.append([0, 0, 0, 0])


def bfs():    

    da = [1, 0, 0, 1, 1, 0, 1]
    db = [0, 1, 0, 1, 0, 1, 1]
    dc = [0, 0, 1, 0, 1, 1, 1]

    while que:
        a, b, c, s = que.popleft()
        for i in range(7):
            na = a+ da[i]
            nb = b+ db[i]
            nc = c+ dc[i]
            ns = s+1
            if na>A or nb > B or nc> C or ns > S:
                continue
            if dp[ns][na][nb][nc] != 0:
                continue

            dp[ns][na][nb][nc] = dp[s][a][b][c] * ns
            que.append((na, nb, nc, ns))

bfs()
```

```python
S, A, B, C = 3, 1, 1, 1
```

```python
dp = [[ [ [-1 for _ in range(C+1)] for _ in range(B+1)] for _ in range(A+1)] for _ in range(S+1)]
da = [-1, 0, 0, -1, -1, 0, -1]
db = [0, -1, 0, -1, 0, -1, -1]
dc = [0, 0, -1, 0, -1, -1, -1]

def dfs(a, b, c, s):
    if s == 0:
        if a+b+c == 0:
            return 1
        else:
            return 0
    if dp[s][a][b][c] != -1:
        return dp[s][a][b][c]
    value = 0
    for i in range(7):
        na = a+ da[i]
        nb = b+ db[i]
        nc = c+ dc[i]
        ns = s-1
        if na<0 or nb < 0 or nc< 0 or ns < 0:
            continue
        value += dfs(na, nb, nc, ns)
    dp[s][a][b][c] = value
    return value

print(dfs(A, B, C, S))
```

```python
dp = [[[[-1 for _ in range(C+1)] for _ in range(B+1)] for _ in range(A+1)] for _ in range(S+1)]

def go(n, a, b, c):
    if n == 0:
        if a == 0 and b == 0 and c == 0:
            return 1
        else:
            return 0
    if a < 0 or b < 0 or c < 0:
        return 0
    if dp[n][a][b][c] != -1:
        return dp[n][a][b][c]
    dp[n][a][b][c] = 0
    for i in range(2):
        for j in range(2):
            for k in range(2):
                if i + j + k == 0:
                    continue
                dp[n][a][b][c] += go(n - 1, a - i, b - j, c - k)
    dp[n][a][b][c] %= 1000000007
    return dp[n][a][b][c]

print(go(S, A, B, C))
```
