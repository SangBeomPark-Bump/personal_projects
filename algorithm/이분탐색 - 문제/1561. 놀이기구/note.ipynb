```python
N, M = 3, 5
arr =  list(map(int, "7 8 9 7 8".split()))
```

```python
def calc(x):
    result = 0
    for i in arr:
        result += (x // i) + 1
    return result

start = 0
end = (N // M) * (max(arr) + 1) 

while start <= end:
    mid = (start + end) // 2
    max_kid = calc(mid)
    if max_kid < N:
        start = mid + 1
    elif max_kid >= N:
        end = mid - 1
        ans_time = mid

ans_time
```

```python
mod =  N - calc(ans_time - 1)

ans = -1
for index,att in enumerate(arr):
    if ans_time % att == 0:
        mod -= 1
    
    if mod == 0:
        ans = index + 1
        break

print(ans)
```
