```python
N = 3

raw_arr = """2 3 3
3 2 3
3 3 2""".split('\n')

arr =  [list(map(int,row.split())) for row in raw_arr]
```

```python
maxmask = (1 << N)
dp = [ [INF for _ in range( maxmask)] for _ in range(N+1) ]
dp[0][0] = 0
maxmask
```

```python
from collections import deque

numbers = [ [] for _ in range(N+1)]
numbers[0].append(0)
numbers

for n in range(1, N+1):
    cur_arr = arr[n-1]
    for i in range(N):
        for mask in numbers[n-1]:
            # if not (mask & (1<<i)):
            dp[n][mask | (1<<i)] = min(dp[n][mask | (1<<i)],  dp[n-1][mask] + cur_arr[i])
    print(dp[n])
print(dp[-1][-1])
```
