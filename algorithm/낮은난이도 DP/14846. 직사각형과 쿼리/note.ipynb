```python
dp = [ [ [0 for _ in range(10)] for _ in range(N+1)] for _ in range(N+1)]
dp
```

```python
for y in range(1, N+1):
    for x in range(1, N+1):
        cur_num = arr[y][x]-1
        for i in range(10):
            dp[y][x][i] = dp[y-1][x][i] + dp[y][x-1][i] - dp[y-1][x-1][i]
        dp[y][x][cur_num] +=1


```

```python
def ans(y1, x1, y2, x2):
    temp = 0
    for i in range(10):
        if (dp[y2][x2][i] - dp[y1-1][x2][i] - dp[y2][x1-1][i] + dp[y1-1][x1-1][i]) > 0:
            if verbose:
                print(f'{i+1}는 존재한다')
            temp+=1
    return temp
```
